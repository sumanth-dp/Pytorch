```python
import torch
import torch.nn as nn
```

```python
class network(nn.Module):

  def __init__(self):

    super().__init__()

    self.model = nn.Sequential(

        nn.Linear(6,8),
        nn.ReLU(),

        nn.Linear(8,4),
        nn.ReLU(),

        nn.Linear(4,1),
        nn.Softmax()
    )

  def forward(self,x):

    output = self.model(x)

    return output
```

```python
x=torch.rand(10,6)
x
```

```python
model=network()
```

```python
y_pred=model(x)
y_pred
```
