```python
import torch
import torch.nn as nn
```

```python
class ANN(nn.Module):
  def __init__(self):

    super().__init__()

    self.layer1 = nn.Linear(5,8)

    self.relu1 = nn.ReLU()

    self.layer2 = nn.Linear(8, 5)

    self.relu2 = nn.ReLU()

    self.output_layer = nn.Linear(5,1)

    self.sigmoid = nn.Sigmoid()

  def forward(self, x):

    output1 = self.layer1(x)

    output2 = self.relu1(output1)

    output3 = self.layer2(output2)

    output4 = self.relu2(output3)

    output5 = self.output_layer(output4)

    output6 = self.sigmoid(output5)


    return output6
```

```python
x=torch.randn(10,5)
print(x)
y=torch.rand(10,1)
print(y)
```

```python
model = ANN()

y_pred = model(x)

y_pred
```
