```python
import torch

class SimpleNN():

    def __init__(self, x, y):
        self.w = torch.rand(x.shape[1], 1,dtype=torch.float64, requires_grad=True)
        self.b = torch.zeros(1, dtype= torch.float64, requires_grad=True)

    def forwardPass(self):
        z = torch.matmul(x,self.w) + self.b
        y_pred = torch.sigmoid(z)
        return y_pred
    # these are retuning nan value
    # def calculate_loss(self, y_pred):
    #     loss = -(self.y * torch.log(y_pred) +
    #              (1 - self.y) * torch.log(1 - y_pred))
    #     return loss.mean()

    def calculate_loss(self, y,y_pred):
      y_pred = torch.clamp(y_pred, 1e-7, 1 - 1e-7)

      loss = -(y * torch.log(y_pred) +(1 - y) * torch.log(1 - y_pred))

      return loss.mean()

    def update(self, loss):

        loss.backward()

        with torch.no_grad():
            self.w -= 0.001 * self.w.grad
            self.b -= 0.001 * self.b.grad

            self.w.grad.zero_()
            self.b.grad.zero_()
```
